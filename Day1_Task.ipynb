```python
#create 3*3 array random floats between 0 and 1.
#print its shape, dtype, and the value of row1 and col2.

import numpy as np
# Create a 3x3 array of random floats between 0 and 1
arr = np.random.rand(3, 3)

# Print array
print("Array:")
print(arr)

# Print shape
print("\nShape:", arr.shape)

# Print data type
print("Data Type:", arr.dtype)

# Print Row 1
print("\nRow 1:")
print(arr[0])

# Print Column 2
print("\nColumn 2:")
print(arr[:, 1])
```

```python
#how many unique values does the embarked column has?
#what is most common passanger class (pclass)?
import pandas as pd

# Load dataset
df = pd.read_excel("/content/titanic_dataset.xlsx")

# Number of unique values in embarked
print("Unique values in embarked:", df["embarked"].nunique())

# Most common passenger class
print("\nPassenger class counts:")
print(df["pclass"].value_counts())

print("\nMost common passenger class:", df["pclass"].mode()[0])
```
