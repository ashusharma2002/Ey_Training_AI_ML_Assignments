```python
#how many unique values does the embarked column has?
#what is most common passanger class (pclass)?
import pandas as pd

# Load dataset
df = pd.read_excel("titanic_dataset.xlsx")

# Number of unique values in embarked
print("Unique values in embarked:", df["embarked"].nunique())

# Most common passenger class
print("\nPassenger class counts:")
print(df["pclass"].value_counts())

print("\nMost common passenger class:", df["pclass"].mode()[0])
```

```python
#using group by find
#Average fare paid by passengers from each embarkation port
#Number of survivors per class sorted from highest to lowest

avg_fare = df.groupby("embarked")["fare"].mean()
print("Average Fare by Embarkation Port:")
print(avg_fare)

# 2. Number of survivors per class sorted from highest to lowest
survivors_per_class = df.groupby("pclass")["survived"].sum().sort_values(ascending=False)

print("\nNumber of Survivors per Class:")
print(survivors_per_class)
```
